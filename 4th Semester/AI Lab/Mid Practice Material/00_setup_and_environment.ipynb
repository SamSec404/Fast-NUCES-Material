## 1. What Is a Virtual Environment?

Imagine you have two projects on your computer:

- **Project A** needs library version 1.0
- **Project B** needs library version 2.0

If both projects share the same Python installation, they'd fight over which version to use. A **virtual environment** solves this by giving each project its own **isolated copy** of Python and its libraries.

Think of it like giving each project its own **private toolbox** — changes in one toolbox don't affect the other.

### Why Use a Virtual Environment?

| Without virtual environment | With virtual environment |
|---|---|
| All projects share the same libraries | Each project has its own libraries |
| Updating one project can break another | Projects are completely isolated |
| Hard to know which libraries a project needs | Easy to track dependencies |
| Messy and error-prone | Clean and reproducible |

## 2. Creating a Virtual Environment

Python comes with a built-in module called `venv` that creates virtual environments. You do this from your **terminal** (not from inside Python).

### Step 1: Open a Terminal

- **Linux/Mac**: Open the Terminal app
- **Windows**: Open Command Prompt or PowerShell

### Step 2: Navigate to Your Project Folder

```bash
cd /path/to/your/project
```

For example:

```bash
cd ~/Desktop/py
```

### Step 3: Create the Virtual Environment

```bash
python3 -m venv venv
```

Let's break that command down:

| Part | Meaning |
|---|---|
| `python3` | Run Python 3 |
| `-m venv` | Use the built-in `venv` module |
| `venv` | Name of the folder to create (you can call it anything, but `venv` is the convention) |

After running this, you'll see a new folder called `venv/` in your project directory. That folder **is** your virtual environment.

### Step 4: Activate the Virtual Environment

Creating the environment isn't enough — you need to **activate** it so your terminal knows to use it.

**Linux / Mac:**
```bash
source venv/bin/activate
```

**Windows (Command Prompt):**
```bash
venv\Scripts\activate
```

**Windows (PowerShell):**
```bash
venv\Scripts\Activate.ps1
```

Once activated, you'll see the environment name in your terminal prompt:

```
(venv) ayyzenn@laptop:~/Desktop/py$
```

That `(venv)` at the beginning tells you the virtual environment is active.

### Deactivating

When you're done working, you can leave the virtual environment:

```bash
deactivate
```

The `(venv)` prefix will disappear, and you're back to your system Python.

## 3. Installing Libraries with `pip`

Python has a massive ecosystem of **libraries** (also called **packages**) — pre-written code that other people have built so you don't have to reinvent the wheel.

`pip` is Python's **package manager**. It downloads and installs libraries from [PyPI](https://pypi.org/) (the Python Package Index — a huge online catalog of packages).

### Basic Commands

Make sure your virtual environment is **activated** first, then:

```bash
# Install a library
pip install library_name

# Install a specific version
pip install library_name==2.1.0

# Upgrade a library to the latest version
pip install --upgrade library_name

# Uninstall a library
pip uninstall library_name

# See all installed libraries
pip list

# Save your project's dependencies to a file
pip freeze > requirements.txt

# Install all dependencies from a file (useful when sharing projects)
pip install -r requirements.txt
```

### Example: Installing the `requests` Library

```bash
(venv) $ pip install requests
```

Output:

## 4. What Is a Jupyter Notebook?

You're reading one right now! A **Jupyter Notebook** (files ending in `.ipynb`) is an interactive document that mixes:

- **Code cells** — Write and run Python code, see results immediately
- **Markdown cells** — Write formatted text, explanations, headings, tables, links, images
- **Output** — Results from code appear right below each cell

This makes notebooks perfect for:
- Learning and experimenting with code
- Data analysis and visualization
- Sharing your work with explanations
- Step-by-step tutorials (like this one!)

### The Name "Jupyter"

The name comes from three programming languages: **Ju**lia, **Pyt**hon, and **R** — but today Jupyter supports many more languages.

### How to Install Jupyter

With your virtual environment activated:

```bash
(venv) $ pip install jupyter
```

This installs the full Jupyter ecosystem. To launch it:

```bash
(venv) $ jupyter notebook
```

This opens a browser window where you can create and edit `.ipynb` files.

## 5. What Is `ipykernel`?

A **kernel** is the engine that actually runs your code. When you run a cell in a notebook, the text goes to the kernel, the kernel executes it, and sends the result back.

`ipykernel` is the Python kernel for Jupyter. It's the bridge between your notebook and a specific Python installation.

### Why Do You Need It?

Here's the common problem:

1. You create a virtual environment and install your libraries in it
2. You open a notebook in VS Code
3. **But the notebook can't find your libraries!**

This happens because the notebook doesn't know about your virtual environment yet. You need to **register** your virtual environment as a kernel that Jupyter can use.

### Step-by-Step: Connecting Your Virtual Environment to Jupyter

**Step 1:** Activate your virtual environment
```bash
source venv/bin/activate  # Linux/Mac
```

**Step 2:** Install `ipykernel` inside the virtual environment
```bash
(venv) $ pip install ipykernel
```

**Step 3:** Register the environment as a Jupyter kernel
```bash
(venv) $ python -m ipykernel install --user --name=my_project --display-name "My Project (Python 3)"
```

### Why do we need to "register" the kernel?

By default, Jupyter and VS Code only know about a few Python installations — usually the system Python or wherever Jupyter was installed. They have **no idea** that your virtual environment exists. Your `venv/` folder is just a directory; nothing is linking it to Jupyter.

When you run `python -m ipykernel install`, you're telling Jupyter: *"Hey, there's another Python over here — the one inside my active virtual environment. Add it to your list of available kernels."*

**What actually happens:**

1. The command creates a **kernel spec** — a small JSON file that stores the **path** to the Python executable in your venv (e.g. `~/Desktop/py/venv/bin/python`) and the **display name** you want in the kernel picker.
2. This spec is saved in your user directory (because of `--user`), so it persists across projects.
3. When you open a notebook and pick "My Project (Python 3)", Jupyter reads that spec and launches a process using *that* Python — the one with all your venv's libraries.

**Without this step:** The notebook would use whatever default kernel it finds (often system Python), and `import requests` or your other installed packages would fail with `ModuleNotFoundError`.

**With this step:** The notebook runs code in the same environment where you ran `pip install` — your venv's Python and its packages.

| Part | Meaning |
|---|---|
| `python -m ipykernel install` | Create and register a kernel spec for the *currently active* Python |
| `--user` | Save the kernel spec in your user folder (no admin rights needed) |
| `--name=my_project` | Internal ID for the kernel (use this when removing it) |
| `--display-name "My Project (Python 3)"` | The friendly name in the kernel picker dropdown |

**Step 4:** In VS Code, click the kernel selector (top-right of the notebook) and choose **"My Project (Python 3)"**

Now your notebook uses the exact same Python and libraries as your virtual environment!

### Managing Kernels

Here are some useful commands for managing your registered kernels:

```bash
# List all available kernels
jupyter kernelspec list

# Remove a kernel you no longer need
jupyter kernelspec remove my_project
```

## 6. Putting It All Together — Full Example

Here's the complete workflow from scratch. Copy and paste these commands into your terminal one by one:

```bash
# 1. Create a project folder and enter it
mkdir ~/Desktop/my_project
cd ~/Desktop/my_project

# 2. Create a virtual environment
python3 -m venv venv

# 3. Activate it
source venv/bin/activate          # Linux/Mac
# venv\Scripts\activate           # Windows

# 4. Upgrade pip (good practice)
pip install --upgrade pip

# 5. Install the tools you need
pip install ipykernel jupyter

# 6. Register the kernel so Jupyter/VS Code can find it
python -m ipykernel install --user --name=my_project --display-name "My Project (Python 3)"

# 7. Install any other libraries you need
pip install requests numpy pandas

# 8. Save your dependencies
pip freeze > requirements.txt

# 9. Open VS Code in this folder
code .
```

That's it! You now have a fully isolated Python environment with Jupyter support.
