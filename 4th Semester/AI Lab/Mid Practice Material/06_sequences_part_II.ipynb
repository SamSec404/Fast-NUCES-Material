## 1. Nested Sequences

Just as we had nested expressions, we can have **nested sequences** — a list whose elements are themselves lists. This is useful for tables, matrices, or any structure where each "row" is a smaller sequence.

### Creating Nested Lists

A list can contain other lists as elements. To access an element in a nested structure, use multiple indices: `pairs[i][j]` means "the element at index `j` of the inner list at index `i`".

```python
digits = [1, 2, 3, 4, 5]

pairs = [
    [1, 20],
    [2, 30],
    [3, 40],
    [4, 50]
]
```

### Iterating with Unpacking

When each inner list has the same length (e.g., pairs of values), you can **unpack** directly in the loop: `for x, y in pairs` assigns the first element of each inner list to `x` and the second to `y`.

```python
pairs[1][0]  # Inner list at index 1, element at index 0 → 2
```

```python
for x, y in pairs:  # Unpack each inner list into x and y
    print("x =", x, "and y =", y)
```

### Example: Counting Matching Pairs and Recording Indices

When you need both the **value** and the **position** (index) while looping, use a separate `index` variable that you increment each iteration. Here we find pairs where both elements are equal and record both the count and where they occur.

```python
pairs = [[1, 1], [2, 2], [2, 3], [4, 4], [8, 8]]
```

```python
same_count = 0
index = 0
same_indices = []

for x, y in pairs:
    if x == y:
        same_count += 1
        same_indices.append(index)
    index += 1

print(same_count)
print(same_indices)
```

## 2. List Methods — Modifying Lists In Place

Lists have built-in **methods** that change the list directly. Unlike `len()` or slicing, these methods modify the list and typically return `None` — so you don't assign the result.

| Method | What it does |
|--------|--------------|
| `list.append(x)` | Add `x` to the end |
| `list.insert(i, x)` | Insert `x` at index `i` (shifts elements right) |
| `list.remove(x)` | Remove the first occurrence of `x` |
| `list.reverse()` | Reverse the list in place |
| `list.sort()` | Sort the list in place |
| `list.index(x)` | Return index of first occurrence of `x` |

### append — Add to the End

`append(x)` adds a single element to the end. No assignment needed — the list is modified in place.

```python
a = [66.25, 333, 333, 1, 1234.5]  # Sample list for method demos
```

```python
print(a)
a.append(987)  # Add 987 to the end
print(a)
```

### insert — Add at a Specific Position

`insert(i, x)` inserts `x` at index `i`, shifting existing elements to the right. If `i` is beyond the list length, the element goes at the end.

```python
a.insert(7, 0)  # Insert 0 at index 7 (shifts existing elements right)
print(a)
```

```python
print(a)
a.insert(2, -1)  # Insert -1 at index 2; elements at 2+ shift right
print(a)
```

```python
a
```

```python
print(a)
a.index(333)  # First occurrence of 333 is at index 1
```

### `index` — Find the Position of a Value

`index(x)` returns the index of the **first** occurrence of `x` in the list. If `x` is not found, it raises `ValueError`, so it's often used with `in` first.

```python
w = [1, 2,1,1,1,1,1,1, 6,5,1,4]
num = 1

if num in w:
    w.remove(num)  # Check with 'in' first; remove raises ValueError if not found

print(w)
```

### `remove` — Remove the First Occurrence

`remove(x)` removes the first occurrence of `x`. It modifies the list in place and returns `None`. If `x` isn't in the list, it raises `ValueError` — check with `in` first if you're not sure.

```python
print(a)

if 333 in a:
    a.remove(333)  # Removes first occurrence only

print(a)
```

```python
print(a)

a.reverse()  # Reverse in place; no assignment needed

print(a)
```

### `reverse` and `sort` — Reorder In Place

- `reverse()` reverses the list order.
- `sort()` sorts the list. Use `sort(reverse=True)` for descending order, `sort(reverse=False)` or `sort()` for ascending.

Both modify the list in place. No assignment needed.

```python
print(a)

a.sort(reverse = True)

print(a)
```

```python
print(a)

a.sort(reverse = False )

print(a)
```

## 3. Strings as Sequences

Strings are **sequences** too — you can index, slice, and iterate over them just like lists. They support `len()`, `in`, and indexing with `[i]` and `[start:stop]`. The key difference: **strings are immutable** — you cannot change a character in place.

```python
# Strings are sequences — index and slice like lists
s = "I am a dummy string. Lorem ipsum."
s[19]  # 20th character (0-based)
```

Strings support the same indexing and slicing as lists. Index `19` gives the 20th character; `s[2:4]` gives the slice from index 2 up to (not including) 4.

```python
len(s)
```

```python
s[2:4]
```

```python
s[1] = 'x'  # Strings are immutable — this raises TypeError!
```

## 4. Ranges

`range(start, stop, step)` generates a sequence of numbers. It includes `start` but stops just before `stop`. The `step` is optional (default 1). Often used in `for` loops.

```python
list(range(1, 10, 3))  # 1, 4, 7 — start 1, stop before 10, step 3
```

```python
for j in range(1, 4):
    print("Printing table for:", j)
    for i in range(1, 3):
        print(j, "×", i, "=", j * i)  # Nested loops: outer j, inner i
```

These are called **nested loops** — one loop inside another.

## 5. List Comprehensions

A **list comprehension** is a concise way to build a list. Syntax: `[expr for item in seq if condition]`. It loops over `seq`, optionally filters by `condition`, and collects `expr` for each item.

```python
# Long form: build evens with a loop
nums = [1, 2, 3, 4, 5, 6, 101, 105, 1000]
evens = []

for i in nums:
    if i % 2 == 0:
        evens.append(i)

print(evens)
```

```python
# Same idea as a comprehension: [expr for item in seq if condition]
multiples_of_5 = [i for i in nums if i % 5 == 0]
print(multiples_of_5)
```

```python
# Divisors of 11: x where 11 % x == 0
n = 11
[x for x in range(1, n + 1) if n % x == 0]
```

## 6. Tuples — Immutable Sequences

**Tuples** use parentheses `()` instead of brackets. They support `len`, indexing, slicing, and iteration — but **tuples are immutable**; you cannot change an element in place.

```python
digits = (1, 8, 2, 8)  # Tuple: parentheses, not brackets
```

```python
len(digits)
```

```python
digits[1]
```

```python
digits[1:3]
```

```python
digits[0] = 11  # Tuples are immutable — this raises TypeError!
```
