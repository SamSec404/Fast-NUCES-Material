## Python vs C++ — Basics

| Concept | Python | C++ |
|---------|--------|-----|
| Class definition | `class Name:` | `class Name { };` |
| Constructor | `def __init__(self, ...):` | `Name(...) { }` |
| Instance reference | `self` (explicit first param) | `this` (implicit) |
| Create object | `obj = MyClass()` | `MyClass obj;` |

## 1. Classes and Objects

A **class** is the blueprint. You create an **object** (instance) by calling the class like a function.

### Defining a Simple Class

**Python:** `class ClassName:` with a colon and indented body. **C++:** `class ClassName { };` with braces.

````python
class Dog:
    pass  # Empty body — like empty { } in C++

# Create an instance — like "Dog rex;" in C++
rex = Dog()
rex
````

`rex` is an **instance** of `Dog`. We can attach attributes to it. (In C++, you'd declare members inside the class.)

````python
rex.name = "Rex"
rex.age = 3
print(rex.name, rex.age)
````

### Multiple Instances — Each Object Is Independent

Create several objects from the same class. Each has its own attributes:

````python
max_dog = Dog()
max_dog.name = "Max"
max_dog.age = 2

print(rex.name, rex.age)   # Rex, 3
print(max_dog.name, max_dog.age)  # Max, 2 — different object, different data
````

---

## 2. The Constructor — `__init__`

`__init__` runs **automatically** when you create an object. It's where you set up the object's initial state — like a constructor in C++. The first parameter is always **`self`**, the instance being created (like `this` in C++).

````python
class Dog:
    def __init__(self, name, age):
        # self = the instance; name and age are the constructor args
        self.name = name   # Like this->name = name in C++
        self.age = age

rex = Dog("Rex", 3)  # __init__ runs automatically here
print(rex.name, rex.age)
````

### Step-by-Step: What Happens When You Call `Dog("Rex", 3)`?

| Step | What Python does |
|------|------------------|
| 1 | Creates a new, empty `Dog` object |
| 2 | Calls `Dog.__init__(that_object, "Rex", 3)` — so `self` is the new object |
| 3 | Inside `__init__`, `self.name = "Rex"` and `self.age = 3` set the attributes |
| 4 | Returns the object (you assign it to `rex`) |

### Default Arguments

You can give default values to constructor parameters. If the caller doesn't provide them, the default is used:

````python
class Dog:
    def __init__(self, name, age=1):  # age defaults to 1
        self.name = name
        self.age = age

puppy = Dog("Max")     # age not given → uses 1
adult = Dog("Rex", 5)  # both given
print(puppy.name, puppy.age)
print(adult.name, adult.age)
````

### Modifying Attributes After Creation

Attributes are mutable. You can change them anytime:

````python
rex = Dog("Rex", 3)
print(rex.age)

rex.age = 4  # Rex had a birthday!
print(rex.age)

rex.breed = "Labrador"  # You can even add new attributes on the fly
print(rex.breed)
````

## 3. Methods — Behavior on Objects

**Methods** are functions defined inside a class. The first parameter is always `self`. When you call `obj.method()`, Python automatically passes `obj` as `self`.

````python
class Dog:
    def __init__(self, name, age=1):
        self.name = name
        self.age = age

    def bark(self):
        # self = the instance; gives access to this object's attributes
        print(self.name, "says woof!")

    def human_age(self):
        return self.age * 7  # Rough "dog years"

rex = Dog("Rex", 3)
rex.bark()           # rex is passed as self automatically
print(rex.human_age())
````

When you call `rex.bark()`, Python does `Dog.bark(rex)` behind the scenes — so `self` is `rex`. That's how the method knows which dog's `name` to use.

### Example: Another Class — `Cat`

Same idea, different data and behavior:

````python
class Cat:
    def __init__(self, name, age=1):
        self.name = name
        self.age = age

    def meow(self):
        print(self.name, "says meow!")

whiskers = Cat("Whiskers", 2)
whiskers.meow()

rex = Dog("Rex", 3)
rex.bark()
````

**C++ equivalent:**

```cpp
class Dog {
public:
    Dog(std::string n, int a) : name(n), age(a) {}
    void bark() { cout << name << " says woof!" << endl; }
    int human_age() { return age * 7; }
private:
    std::string name;
    int age;
};
rex.bark();  // 'rex' is implicit 'this'
```

## 4. Common Errors — For Understanding

These cells raise errors **on purpose** so you see what happens when things go wrong.

### Error 1: Missing Constructor Arguments

`Dog` expects `name` and `age`. If you call `Dog()` with no arguments, Python raises **`TypeError`**:

````python
Dog()  # __init__ requires name and age — TypeError (intentional)
````

### Error 2: Accessing an Attribute That Doesn't Exist

If you never set `rex.breed` (and `__init__` didn't set it), accessing it raises **`AttributeError`**:

````python
rex = Dog("Rex", 3)
rex.breed  # breed was never set — AttributeError (intentional)
````

### Error 3: Calling a Method Without an Instance

`bark` expects `self` (the instance). If you call `Dog.bark()` with no object, Python has nothing to pass as `self` — **`TypeError`**:

````python
Dog.bark()  # Who is 'self'? No instance! — TypeError (intentional)
````

The **correct** way: create an object first, then call the method on it: `rex = Dog("Rex", 3)` and `rex.bark()`.

## 5. Example: A Simple `Rectangle` Class

Putting it all together — attributes, constructor, and a method that uses them:

````python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

r = Rectangle(4, 5)
print("Area:", r.area())
print("Perimeter:", r.perimeter())
````

## Quick Reference

| Concept | Python | C++ |
|---------|--------|-----|
| Define class | `class X:` | `class X { };` |
| Constructor | `def __init__(self, a, b):` | `X(int a, int b) { }` |
| Instance ref | `self` | `this` |
| Set attribute | `self.x = value` | `this->x = value;` |
| Method | `def m(self):` | `void m() { }` |
| Create object | `obj = X(a, b)` | `X obj(a, b);` |

| Error | Cause |
|-------|-------|
| `TypeError` | Missing constructor args; or calling method without instance |
| `AttributeError` | Accessing attribute that was never set |
