## 2. Data Types — What Kind of Value Is This?

Every value in Python has a **type**. The most common ones you'll see right away are:

- **`int`** — whole numbers (e.g., `1`, `-5`, `42`)
- **`float`** — decimal numbers (e.g., `3.14`, `0.5`, `-2.7`)
- **`str`** — text/strings (e.g., `"hello"`, `'world'`)
- **`bool`** — True or False

You can check a value's type using the built-in `type()` function:

````python
type(1)  # A whole number is an "int" (integer)
````

````python
type(1 / 2)  # Division always produces a "float" (decimal number)
````

You can mix `int` and `float` values in math. Python will give you a `float` result whenever a decimal is involved:

````python
1/2.0 + 1/4.0  # 0.5 + 0.25 = 0.75
````

````python
2.1 + 5.3  # Adding two floats together
````

## 3. Operator Precedence — Order of Operations

Just like in regular math, Python follows an **order of operations** (sometimes called PEMDAS/BODMAS). Multiplication and division happen **before** addition and subtraction.

Use **parentheses** `()` to force a specific order:

````python
2 + 2 * 2  # Multiplication first: 2 + (2*2) = 2 + 4 = 6
````

````python
(2 + 2) * 2  # Parentheses first: (2+2) * 2 = 4 * 2 = 8
````

> **Want to learn more?** See the full list of [operator precedence rules in the official Python docs](https://docs.python.org/3/reference/expressions.html#operator-precedence).

## 4. Comparison Operators — Asking Yes/No Questions

Comparisons let you ask questions like "Is 3 less than 4?" Python answers with `True` or `False` (these are called **boolean** values).

| Operator | Meaning                  | Example  | Result  |
|----------|--------------------------|----------|---------|
| `<`      | Less than                | `3 < 4`  | `True`  |
| `>`      | Greater than             | `3 > 4`  | `False` |
| `<=`     | Less than or equal to    | `3 <= 3` | `True`  |
| `>=`     | Greater than or equal to | `3 >= 3` | `True`  |
| `==`     | Equal to (comparison)    | `3 == 3` | `True`  |
| `!=`     | Not equal to             | `3 != 4` | `True`  |

**Important:** `==` (double equals) *checks* equality. A single `=` is used for something else (assigning values), which we'll see later.

````python
3 < 4  # Is 3 less than 4? Yes!
````

````python
3 == 4  # Is 3 equal to 4? No!
````

````python
3 == 3  # Is 3 equal to 3? Yes!
````

**What happens if you use `=` instead of `==`?** You get an error! This is one of the most common beginner mistakes. The cell below shows what that looks like — don't worry about the error message details for now:

````python
3 = 3  # This causes an error! Use == for comparison, not =
````

````python
3 > 4  # Is 3 greater than 4? No!
````

````python
3 >= 3  # Is 3 greater than or equal to 3? Yes!
````

````python
2 <= 1  # Is 2 less than or equal to 1? No!
````

## 6. Booleans — True or False

**Booleans** are the simplest type — they can only be one of two values: `True` or `False`.

You can combine booleans using **logical operators**:

- **`and`** — Both sides must be `True` for the result to be `True`
- **`or`** — At least one side must be `True` for the result to be `True`

Think of it like this:
- `and` = "Are **both** things true?"
- `or` = "Is **at least one** thing true?"

### The `and` Operator

````python
True  # A boolean value — just True by itself
````

````python
False  # The other boolean value
````

````python
True and True  #Both are True, so the result is True
````

````python
True and False  # One is False, so "and" gives False
````

````python
False and True  # One is False, so "and" gives False
````

````python
False and False  # Both are False, so definitely False
````

### The `or` Operator

With `or`, the result is `True` if **at least one** side is `True`. The only way to get `False` is if **both** sides are `False`:

````python
True or True  # At least one is True, so result is True
````

````python
True or False  # At least one is True, so result is True
````

````python
False or True  # At least one is True, so result is True
````

````python
False or False  # Neither is True, so result is False
````

### Precedence with `and` / `or`

Just like `*` happens before `+` in math, **`and` is evaluated before `or`** in boolean logic. Parentheses can change this order — compare these two:

````python
# "and" runs first: False and False = False
# Then: True or False or False = True
True or False and False or False
````

````python
# Parentheses change the order:
# (True or False) = True, (False or False) = False
# True and False = False
(True or False) and (False or False)
````

## 7. Call Expressions — Using Built-in Functions

Python comes with many useful **functions** built right in. A function is like a little machine: you give it some input (called **arguments**), and it gives you back a result.

You "call" a function by writing its name followed by parentheses with the arguments inside:

```
function_name(argument1, argument2, ...)
```

Here are some common built-in functions:

| Function | What it does                        | Example          | Result |
|----------|-------------------------------------|------------------|--------|
| `max()`  | Returns the largest value           | `max(3, 7)`      | `7`    |
| `min()`  | Returns the smallest value          | `min(3, 7)`      | `3`    |
| `pow()`  | Raises a number to a power          | `pow(2, 3)`      | `8`    |
| `abs()`  | Returns the absolute (positive) value | `abs(-5)`      | `5`    |
| `type()` | Tells you the type of a value       | `type(42)`       | `int`  |

````python
max(7, 9)  # Which is bigger, 7 or 9?
````

````python
max(2 + 25, 9)  # Python evaluates 2+25=27 first, then compares 27 vs 9
````

````python
pow(3, 2)  # 3 raised to the power of 2 = 3 * 3 = 9
````

### Nesting Function Calls

You can put function calls **inside** other function calls. Python works from the **inside out**, just like nested parentheses in math:

````python
# Step by step:
#   min(1, -2) = -2
#   pow(3, 5) = 243
#   min(243, -4) = -4
#   max(-2, -4) = -2
max(min(1, -2), min(pow(3, 5), -4))                  
````

## 8. A Sneak Peek — Lists and Indexing

Here's a quick preview of something you'll learn more about later: **lists** and **tuples**. A list is a collection of items stored in order. You can access individual items using square brackets `[]` with an **index** (position number, starting from 0):

````python
# Create a list of tuples (tuples are like mini-lists wrapped in parentheses)
a = [(4, 'a'), (3, 'b', 5), ('c', 5, 7)]

# a[2] gets the 3rd item (index starts at 0): ('c', 5, 7)
# a[2][0] gets the 1st element of that tuple: 'c'
type(a[2][0])  # 'c' is a string, so the type is str
````

````python
type(1 / 2)  # A quick reminder: division always gives a float
````
