# Dictionaries — Key–Value Mappings

Lists and tuples store values by **position** (index 0, 1, 2, …). A **dictionary** stores values by **key** — you look up a key and get its value. Think of a real dictionary: you look up a word (key) to find its definition (value).

In this notebook, you'll learn:

1. **Creating dictionaries** — `{key: value, ...}` syntax
2. **Accessing and modifying** — Get, add, and change values
3. **Iteration** — Looping over keys, values, or key–value pairs
4. **Membership** — Checking if a key exists
5. **Practical uses** — Configuration, mapping numbers to names, and type casting

---

## 1. What Is a Dictionary?

A dictionary maps **keys** to **values**. Keys must be unique and immutable (strings, numbers, tuples); values can be anything.

### Creating a Dictionary

Use curly braces `{}` with `key: value` pairs. Keys can be strings, numbers, etc.; values can be any type.

```python
# Recall: lists use integer indices
lst = [1, 2, 3, 4]
lst[0]  # First element by position
```

| Operation | Syntax | Example |
|-----------|--------|---------|
| Get value | `d[key]` | `d['random']` |
| Set/Add | `d[key] = value` | `d['variable'] = "xy"` |
| Membership | `key in d` | `'package' in d` → `True` |

```python
d = {
    "package": "A package is a module that can be added to any program",
    "random": "Any data or information that has no order",
    "unary operator": "An operator that takes only one value for its operation"
} 
```

```python
d['random']  # Get value for key 'random'
```

```python
d['variable'] = "xy"  # Add new key–value pair (or update if key exists)
```

```python
d  # Display the whole dictionary
```

```python
d['python']  # Key not in dict — raises KeyError (intentional, to show the error)
```

---

## 2. Iterating Over Dictionaries

Looping over a dict gives you the **keys**. Use `d[k]` to get each value.

```python
for k in d:  # Iterates over keys
    v = d[k]
    print(k, "=", v)
```

```python
'package' in d  # Check if key exists
```

```python
'else' in d  # Not in dict → False
```

```python
if 'else' not in d:  # Check before accessing to avoid KeyError
    print("The keyword 'else' is not in d... how can this be?")
else:
    print("This keyword was found.")
```

```python
list(d.items())  # List of (key, value) tuples — like [[1,2], [2,2]] but as pairs 
```

### Iterating over key–value pairs

`d.items()` gives `(key, value)` pairs. Unpack in the loop just like with `[[1,2], [2,2]]`.

```python
for k, v in d.items():  # Unpack (key, value) directly; list() optional
    print(k, "=", v)
```

## 3. Using Dicts for Configuration

Dictionaries are ideal for settings or config: named keys instead of positional indices. 

### Keys can be numbers too

Dict keys aren't limited to strings — integers work for mappings like number → name.

```python
autograder_conf = {
    'url': '121.52.146.108:8000',
    'username': 'p1111111',
    'submission_pass': 'abc-334'
}

print('Connecting to', autograder_conf['url'])
```

```python
num_names = {
    1: "one",
    2: "two",
    3: "three",
    20: "twenty",
    30: "thirty",
}
```

```python
num_names[30]  # Key is int 30, not string "30"
```

```python
v = '1'
num_names[v]  # KeyError: keys are ints, '1' ≠ 1 (intentional)
```

```python
num_names[int(v)]  # Cast string to int to match the key type
```

This is called **type casting** — converting `'1'` (str) to `1` (int) to match the dict's key type.

## Quick Reference

| Operation | Syntax | Example |
|-----------|--------|----------|
| Create | `{k: v, ...}` | `d = {'a': 1}` |
| Get | `d[key]` | `d['a']` → `1` |
| Set/Add | `d[key] = value` | `d['b'] = 2` |
| Membership | `key in d` | `'a' in d` → `True` |
| Iterate keys | `for k in d:` | Loop over keys |
| Iterate pairs | `for k, v in d.items():` | Unpack (key, value) |
