## What Is a Sequence?

A **sequence** is an ordered collection of values. "Ordered" means each item has a position (first, second, third, …), so you can access elements by their position.

Python has several sequence types. They're not all the same *type*, but they share common behavior:

| Operation | Meaning | Example |
|---|---|---|
| `len(seq)` | Number of elements | `len([1, 2, 3])` → `3` |
| `seq[i]` | Get element at position `i` | `[1, 2, 3][1]` → `2` |
| `x in seq` | Is `x` in the sequence? | `2 in [1, 2, 3]` → `True` |

The most important sequence for beginners is the **list**.

## 1. Lists — Ordered Collections of Anything

A **list** is a sequence enclosed in square brackets `[]`. It can hold any mix of values — numbers, strings, or even other lists. Lists can grow to any length.

### Creating a List

Separate items with commas. Spaces and line breaks are optional (they just affect readability): 

````python
digits = [1, 8, 2, 8]  # A list of four numbers
````

````python
digits  # Display the whole list
````

````python
len(digits)  # How many elements? 4
````

### Accessing Elements by Index

Use **square brackets** with an **index** (position number) to get one element. Python uses **0-based indexing** — the first element is at index `0`, the second at index `1`, and so on.

| Index | `digits` | Value |
|---|---|---|
| 0 | first | 1 |
| 1 | second | 8 |
| 2 | third | 2 |
| 3 | fourth | 8 |

So `digits[0]` is `1`, and `digits[3]` is `8`.

````python
v = 0
digits[v + 1]  # digits[1] → the second element (index 1)
````

### What Happens with an Invalid Index?

If you use an index that doesn't exist (too small or too large), Python raises an **`IndexError`**:

````python
# Index 4 doesn't exist — list only has indices 0–3!

digits[4]
````

### Membership — Is a Value in the List?

The `in` operator checks whether a value appears *anywhere* in the list:

````python
digits  # Our list: [1, 8, 2, 8]
````

````python
1 in digits  # Is 1 in the list? Yes!
````

````python
88 in digits  # Is 88 in the list? No!
````

## 2. Iterating Over Lists

You can loop over a list with a `for` loop — each iteration gives you the next element. This is exactly like the tea example from the repetition notebook, but now we're using a **list** instead of a tuple:

Because lists are sequences, we can pass them directly to a `for` loop:

````python
def make_tea(ing):
    for i in ing:
        print("Put", i, "in.")

ingredients = ['water', 'milk', 'cinnamon', 'tea leaves']
make_tea(ingredients)
````

### Counting Occurrences

Here's a common pattern: loop through a list and count how many times a value appears. We'll write a `count` function:

````python
seq = [1, 2, 8, 4, 6, 8, 9, 15]  # 8 appears twice
````

````python
def count(s, value):
    """Count how many times value occurs in s."""
    counter = 0
    for i in s:
        if i == value:
            counter = counter + 1
    return counter
````

````python
count(seq, 88)  # 88 doesn't appear — returns 0
````

````python
count(seq, 8)  # 8 appears twice — returns 2
````

## 3. Modifying Lists — Changing Values In Place

Unlike strings, **lists are mutable** — you can change their elements without creating a new list. Use assignment with an index:

```python
list[index] = new_value
```

````python
digits = [1, 2, 3, 4]
````

````python
digits[1]  # Currently 2
````

````python
digits[1] = 0  # Replace the element at index 1
````

````python
digits  # The list is changed — index 1 is now 0
````

## 4. Slicing — Getting a Range of Elements

**Slicing** lets you get a **sub-list** (a range of elements) using the syntax `list[start:stop]`. Important: the slice goes **from `start` up to but not including `stop`**.

| Syntax | Meaning | Example |
|---|---|---|
| `list[start:stop]` | Elements from index `start` to `stop-1` | `digits[1:4]` → indices 1, 2, 3 |
| `list[start:]` | From `start` to the end | `digits[2:]` → indices 2 onwards |
| `list[:stop]` | From the beginning to `stop-1` | `digits[:3]` → indices 0, 1, 2 |
| `list[-4:-1]` | Negative indices count from the end | -1 = last, -2 = second-to-last |

````python
digits = [1, 5, 4, 6, 7, 9]  # indices: 0  1  2  3  4  5
````

````python
digits[1:4]  # Indices 1, 2, 3 — from index 1 up to (but not including) 4
````

````python
len(digits)  # 6 elements
````

### Negative Indices

Python lets you count from the **end** using negative indices: `-1` is the last element, `-2` is the second-to-last, and so on:

````python
digits[-4:-1]  # From 4th-from-end up to (not including) the last element
````

### Omitting start or stop

Leave `start` empty to begin at the start; leave `stop` empty to go to the end:

````python
digits[2:]  # From index 2 to the end — omitting stop means "rest of list"
````

## Quick Reference

### List Operations

| Operation | Syntax | Example |
|---|---|---|
| Create | `[item1, item2, ...]` | `[1, 2, 3]` |
| Length | `len(list)` | `len([1, 2, 3])` → `3` |
| Get element | `list[i]` | `nums[0]` → first element |
| Get slice | `list[start:stop]` | `nums[1:3]` → elements 1 and 2 |
| Modify element | `list[i] = value` | `nums[0] = 99` |
| Membership | `x in list` | `2 in [1, 2, 3]` → `True` |
| Loop | `for x in list:` | Iterate over each element |

### Indexing reminder

- **0-based**: First element is index `0`
- **Negative indices**: `-1` = last, `-2` = second-to-last
- **Slicing**: `list[a:b]` gives elements from index `a` up to (but not including) `b`
