# Library Functions & Variables

In the previous notebook, we used built-in functions like `max()`, `min()`, and `type()`. But Python also comes with a huge collection of extra tools organized into **libraries** (also called **modules**).

In this notebook, you'll learn:

1. **Importing libraries** — How to access extra functions like `sqrt()`
2. **Variables** — How to give names to values so you can reuse them
3. **Importing variables** — Libraries can also provide useful constants like `pi`
4. **Functions as values** — Functions themselves can be stored in variables
5. **Side effects & `print()`** — The difference between returning a value and displaying it

---

## 1. Importing Library Functions

A **library** is a collection of pre-written functions that someone else has already built. Python's **`math`** library, for example, gives you functions for square roots, trigonometry, logarithms, and more.

To use a library, you first need to **import** it. There are two main ways:

### Way 1: Import the Entire Library

When you `import math`, you get access to *everything* in the math library. But you have to use the library name as a prefix (e.g., `math.sqrt()`):

````python
import math  # Import the entire math library
````

````python
math.sqrt(36)  # "math dot sqrt" — square root of 36
````

### Way 2: Import a Specific Function

If you only need one function, you can import just that function using `from ... import ...`. This way you don't need the `math.` prefix:

````python
from math import sqrt  # Import just the sqrt function from math
````

````python
sqrt(256)  # No prefix needed — we imported sqrt directly
````

### Exploring a Library with `help()`

Not sure what's inside a library? Use the built-in `help()` function to see all the functions and constants it provides. (The output is long — scroll through it!)

````python
help(math)  # Shows documentation for everything in the math library
````

## 2. Variables — Giving Names to Values

So far, we've been typing numbers and expressions directly. But what if you want to **reuse** a value later? That's what **variables** are for.

A variable is just a **name** that points to a value. You create one using the **`=`** sign (called the **assignment operator**):

```python
name = value
```

This is **not** the same as the math equals sign! In Python, `=` means **"store this value under this name"**.

````python
x = 24  # Store the value 24 under the name "x"
````

````python
x  # Now just typing "x" gives us back the value we stored
````

You can **reassign** a variable at any time — the old value is simply replaced:

````python
x = 3  # x used to be 24, now it's 3
````

````python
x * 2  # Uses the current value of x (3), so 3 * 2 = 6
````

You can also store the **result of an expression** in a variable:

````python
y = x * 2  # Python evaluates x * 2 first (= 6), then stores it in y
````

````python
y  # y is now 6
````

````python
type(y)  # y holds 6, which is an integer
````

````python
y * 3  # We can use y in expressions just like a number: 6 * 3 = 18
````

> **Fun fact:** In languages like C or Java, you have to declare a variable's type (e.g., `int x = 24;`). In Python, you don't — Python figures out the type automatically. This is called **dynamic typing**.

## 3. Importing Variables from Libraries

Libraries don't just provide functions — they can also provide useful **constants** (values that never change). The `math` library, for example, provides the mathematical constant **pi** (3.14159...):

````python
from math import pi  # Import the constant pi from the math library
````

````python
pi  # The value of pi, accurate to many decimal places
````

Now let's use `pi` and variables together to calculate the **area of a circle** (formula: area = pi x radius²):

````python
pi * 4 * 4  # Area of a circle with radius 4: pi * 4 * 4
````

That works, but it's much more **readable** when we use variables with meaningful names:

````python
radius = 4  # Clear variable name tells us what this number represents
````

````python
pi * pow(radius, 2)  # pi * radius² — much easier to understand!
````

````python
# You can use any variable names, but descriptive names are better practice
a = 4
b = 2
pi * pow(a, b)  # Same result, but "a" and "b" don't tell us what they mean
````

### What Happens If You Use a Variable That Doesn't Exist?

If you try to use a name that Python doesn't recognize (one you haven't assigned a value to), you'll get a **`NameError`**. This is Python's way of saying "I have no idea what that is!"

````python
v  # We never assigned a value to "v", so this will cause an error!
````

> **Takeaway:** Always make sure you've **defined** a variable (assigned it a value) before trying to use it. If you see a `NameError`, check for typos or run the cell that creates the variable first.

## 4. Functions Are Values Too!

Here's a surprising idea: in Python, **functions are values** — just like numbers and strings. A function has a name, and that name is really just a variable that points to the function.

This means you can:
- Look at what a function name refers to
- Assign a function to a **new name**
- Call the function using either name

Let's see this in action. First, what happens if we type a name that hasn't been defined?

````python
f  # "f" isn't defined yet — this will cause a NameError
````

Now, `sqrt` is a function we already imported. We can call it normally, but we can also just look at it without calling it (no parentheses):

````python
sqrt(25)  # Calling sqrt with parentheses — gives us the result
````

````python
sqrt  # Without parentheses — shows us the function object itself
````

Since functions are values, we can **assign a function to a new variable**. Now both names point to the same function:

````python
f = sqrt  # "f" now points to the same function as "sqrt"
````

````python
sqrt(25)  # Calling the original name
````

````python
f(25)  # Calling our new name — same result!
````

## 5. Side Effects & the `print()` Function

So far, we've been seeing output because notebooks automatically **display the result of the last expression** in a cell. But what if you want to display something *on purpose*, or display *multiple* things?

That's what **`print()`** does. It's a built-in function that **displays a value on the screen**.

### Return Value vs. Side Effect

This is an important distinction:

- A **return value** is the result that a function gives back (e.g., `sqrt(25)` returns `5.0`)
- A **side effect** is something a function does *besides* returning a value — like displaying text on screen

`print()` is the classic example of a side effect: it displays something, but it doesn't actually return a useful value.

````python
print(5.0)  # print() displays the value to the screen
````

````python
"a"  # Without print(), the notebook just shows the raw value (with quotes)
````

You can pass **multiple values** to `print()`, separated by commas. Python will display them with spaces in between:

````python
print("The square root of 25 is:", sqrt(25))  # Mix text and computed values
````

### `print()` vs. Notebook Display — A Key Difference

When you view a string *without* `print()`, the notebook shows it with quotes and escape characters (like `\n` for new lines). When you use `print()`, it shows the string as it would actually look:

````python
s = """Multi-line string.
Something on the second line."""
````

````python
s  # Without print: shows raw representation with \n for the line break
````

````python
print(s)  # With print: displays the text properly on two lines
````

## Quick Reference

| Concept | Syntax | Example |
|---|---|---|
| Import entire library | `import library` | `import math` |
| Use function from library | `library.function()` | `math.sqrt(36)` |
| Import specific function | `from library import func` | `from math import sqrt` |
| Import a constant | `from library import name` | `from math import pi` |
| Get help on a library | `help(library)` | `help(math)` |
| Create a variable | `name = value` | `x = 42` |
| Reassign a variable | `name = new_value` | `x = 10` |
| Check a type | `type(value)` | `type(x)` |
| Print output | `print(value)` | `print("Hello!")` |
| Assign function to variable | `new_name = function` | `f = sqrt` |
