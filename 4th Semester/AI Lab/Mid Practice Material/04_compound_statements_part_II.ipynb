# Compound Statements (Part II) — Putting It All Together

In Part I we learned about `if` / `elif` / `else`. Now let's put **everything** we've learned so far — functions, conditionals, and recursion — to work on a real problem.

We'll build a **square root calculator** from scratch using a simple algorithm. This is a great exercise because it combines:

- **Functions** — breaking the problem into smaller pieces
- **Conditionals** — handling different cases
- **Recursion** — a function calling *itself* to gradually improve an answer

---

## 1. The Square Root Algorithm

Here's the idea in plain English:

1. Start with a **guess** for the square root of `x`
2. Check if the guess is **good enough** (is `guess²` close to `x`?)
3. If yes → we're done, return the guess
4. If no → **improve** the guess and try again (go back to step 2)

This is a form of **recursion** — the function calls itself with a better guess each time, getting closer and closer to the real answer.

Let's build this step by step. We need three functions:

| Function | Job |
|---|---|
| `sqrt(x, guess)` | The main function — coordinates the whole process |
| `good_enough(guess, x)` | Checks if our guess is close enough |
| `improve_guess(guess, x)` | Makes the guess a little better |

### The Main Function

````python
def sqrt(x, guess=0.0):
    if x < 0:                          # Can't take sqrt of a negative number
        return None

    if good_enough(guess, x):          # Is our guess close enough?
        return guess                   # Yes — we're done!
    else:
        new_guess = improve_guess(guess, x)  # Make the guess a little better
        return sqrt(x, new_guess)            # Try again with the improved guess (recursion!)
````

### The "Good Enough" Check

How do we know if a guess is close enough? We square the guess and see if it's **within 0.1** of the target number `x`. If `guess² ≈ x`, we're done:

````python
def good_enough(guess, x):
    # If |guess² - x| is less than 0.1, the guess is close enough
    if abs(guess * guess - x) < 0.1:
        return True
    else:
        return False
````

### Improving the Guess

This is the simplest possible approach: if the guess is too low, nudge it up by `0.1`; if it's too high, nudge it down by `0.1`:

````python
def improve_guess(guess, x):
    if guess < x:            # Guess is too small
        return guess + 0.1   # Nudge it up
    else:                    # Guess is too big
        return guess - 0.1   # Nudge it down
````

---

## 2. Testing Our Square Root Function

Now that all three functions are defined, let's try them out. We'll start with a guess close to the answer to keep the example short:

````python
print(sqrt(4, 1.8))  # sqrt(4) should be ~2.0, starting from guess = 1.8
````

Here's what happened inside that call:

```
sqrt(4, 1.8)
  → good_enough(1.8, 4)? → |1.8² - 4| = |3.24 - 4| = 0.76 → No
  → improve_guess(1.8, 4) → 1.8 < 4, so return 1.9
  → sqrt(4, 1.9)
    → good_enough(1.9, 4)? → |1.9² - 4| = |3.61 - 4| = 0.39 → No
    → improve_guess(1.9, 4) → 1.9 < 4, so return 2.0
    → sqrt(4, 2.0)
      → good_enough(2.0, 4)? → |2.0² - 4| = |4.0 - 4| = 0.0 → Yes!
      → return 2.0
```

Each recursive call gets one step closer until the guess is good enough.

### Handling Edge Cases

What happens if we pass a **negative** number? There's no real square root for negatives, so our function returns `None`:

````python
print(sqrt(-2))  # Negative input → the first "if" catches it and returns None
````

## 3. What We Learned

Let's recap the big ideas from this example:

### Recursion — Functions That Call Themselves

```python
def sqrt(x, guess=0.0):
    ...
    return sqrt(x, new_guess)  # ← calls itself!
```

A **recursive function** calls itself with updated arguments. Each call gets closer to the answer. This works because:
- There's a **base case** that stops the recursion (`good_enough` returns `True`)
- Each call makes **progress** toward the base case (the guess improves)

> **Warning:** If there's no base case, or the function never reaches it, you'll get an infinite loop (Python will eventually crash with a `RecursionError`).

### Decomposition — Breaking a Problem into Pieces

Instead of writing one giant function, we split the problem into three small, focused functions:

| Function | Single responsibility |
|---|---|
| `sqrt` | Coordinates the process |
| `good_enough` | Decides when to stop |
| `improve_guess` | Decides how to get closer |

This makes the code easier to **read**, **test**, and **change**. If you wanted a better improvement strategy, you'd only need to modify `improve_guess` — the other functions stay the same.
