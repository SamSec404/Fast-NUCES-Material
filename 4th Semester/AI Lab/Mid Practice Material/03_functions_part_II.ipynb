## 1. Lexical Scope — Where Can You See a Variable?

When you create a variable, it doesn't exist *everywhere* in your program. **Scope** is the word we use to describe *where* a variable is visible.

Python has two main scopes to know about:

| Scope | Where the variable is created | Who can see it? |
|---|---|---|
| **Local** | Inside a function | Only that function |
| **Global** | Outside any function (at the top level) | The whole program |

### The Rules

1. Variables created **inside** a function are **local** — they disappear when the function finishes
2. Variables created **outside** all functions are **global** — functions can *read* them
3. If a local and global variable have the **same name**, the function uses the local one

Let's see this in action:

Pay close attention to the comments in the code below — they label each variable as **local**, **global**, or **parameter**:

````python
def f(x):          # x is a "parameter" — local to this function
    y = 1          # y is a "local variable" — only exists inside f
    x = x + y      # this x is local, not the global one
    print('. x =', x)
    print('. p =', p)  # p is not defined here, so Python looks for a global p
    return x
````

````python
p = 10  # Global variable — visible everywhere
x = 3   # Global variable — but f() has its own local x

z = f(x)  # Calls f with x=3

print('x =', x)  # Still 3! The global x was never changed
print('y =', y)   # Error! y was local to f — it doesn't exist out here
````

Let's walk through what happened step by step:

1. We set `p = 10` and `x = 3` at the **global** level
2. We call `f(3)` — inside the function, a **local** `x` is created with value `3`
3. `y = 1` creates a **local** variable `y`
4. `x = x + y` changes the **local** `x` to `4` (the global `x` is untouched!)
5. `print('. p =', p)` works because `p` is a **global** variable — the function can read it
6. Back outside, `print('x =', x)` shows `3` — the global `x` never changed
7. `print('y =', y)` fails with a **NameError** because `y` was local to `f` and no longer exists

> **Key insight:** Modifying `x` inside a function creates or updates a **local** copy. The original global `x` stays the same.

### A Tricky Case — Assignment Inside a Function

Here's a subtle gotcha that trips up many beginners:

````python
y = 25  # Global y

def f():
    print('y =', y)  # Tries to read y...
    y = 1             # ...but this line makes y LOCAL to the whole function!

f()  # Error! Python sees the assignment below and treats y as local from the start
````

**What happened?** This is an `UnboundLocalError`. Python scans the entire function before running it. When it sees `y = 1`, it decides that `y` is a **local** variable. But then `print('y =', y)` tries to read `y` *before* the assignment, so Python complains that the local `y` doesn't have a value yet.

> **Rule of thumb:** If you assign to a variable *anywhere* inside a function, Python treats that variable as local to the *entire* function — even lines before the assignment.

## 2. Designing Good Functions

Writing functions that *work* is one thing. Writing functions that are **clean, readable, and reusable** is another — and it's a skill worth developing early.

Here are the key principles:

### Use Meaningful Names

Your function name should describe **what** it does. Parameter names should describe **what** they represent.

```python
# Bad — what does this do? What are a, b, c?
def f(a, b, c):
    return a * b * c

# Good — immediately clear
def calculate_volume(length, width, height):
    return length * width * height
```

### Keep Functions Focused

Each function should do **one thing** well. If your function is getting long or doing multiple unrelated tasks, split it up.

### Minimize Dependence on Global Variables

Functions should get their data through **parameters**, not by reaching out to global variables. This makes them more predictable and easier to reuse.

### Document Your Functions with Docstrings

A **docstring** is a special string at the very beginning of a function that explains what it does. It goes inside triple quotes (`""" ... """`):

Here's a well-documented function using the ideal gas law. Notice how the docstring explains the purpose, the formula, and what each parameter means:

````python
def pressure(v, t, n):
    """Compute the pressure in pascals of an ideal gas.

    Applies the ideal gas law: P = n * k * T / V

    v -- volume of gas, in cubic meters
    t -- absolute temperature in degrees kelvin
    n -- particles of gas
    """
    k = 1.38e-23  # Boltzmann's constant
    return n * k * t / v
````

The best part? When you (or anyone else) uses `help()` on your function, the docstring is what they see:

````python
help(pressure)  # Shows the docstring we wrote — this is how documentation works!
````

````python
Let's also use our function to compute something:
````

````python
# Pressure of 1 mole of gas (6.022e23 particles) at room temp (298 K) in 1 cubic meter
pressure(1, 298, 6.022e23)
````

## Quick Reference

### Scope Rules

| Rule | Example |
|---|---|
| Variables inside a function are **local** | `y = 1` inside `def f()` — `y` doesn't exist outside |
| Functions can **read** global variables | `print(p)` inside `f()` works if `p` is global |
| Assignment inside a function makes a variable **local** | `x = 5` inside `f()` — creates a local `x`, doesn't touch global `x` |
| Assigning anywhere in a function makes it local **everywhere** in that function | `y = 1` at the bottom makes `y` local even at the top |

### Good Function Design

| Principle | Why |
|---|---|
| Use meaningful names | `calculate_area(radius)` beats `f(r)` |
| Keep functions focused | One function, one job |
| Use parameters, not globals | Makes functions predictable and reusable |
| Write docstrings | Future you (and others) will thank you |

### Docstring Template

```python
def my_function(param1, param2):
    """Brief description of what the function does.

    param1 -- description of first parameter
    param2 -- description of second parameter
    """
    # function body
    return result
```
