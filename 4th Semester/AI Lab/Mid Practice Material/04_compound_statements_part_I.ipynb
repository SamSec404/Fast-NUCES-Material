This is where **compound statements** come in. In this notebook, you'll learn:

1. **What compound statements are** — The general structure Python uses for multi-line logic
2. **Conditional statements (`if` / `elif` / `else`)** — How to make your code choose between different paths

## 1. What Is a Compound Statement?

A **compound statement** is any Python statement that contains other statements inside it. You've already seen one — `def` (function definitions)!

They all follow the same general pattern:

```python
<header>:
    <statement>
    <statement>
    ...
<separating header>:
    <statement>
    <statement>
    ...
```

The **header** line ends with a colon (`:`), and the **body** lines are **indented** underneath it. Python uses this indentation to know which lines belong together.

### Functions Are Compound Statements

You already know one compound statement — `def`. The header is the `def` line, and the body is everything indented underneath:

Here's a function that calculates the percentage difference between two numbers:

````python
def percent_difference(x, y):  # ← Header (ends with a colon)
    difference = abs(x - y)   # ← Body (indented by 4 spaces)
    return 100 * difference / x

result = percent_difference(40, 50)  # Call the function
result
````

Notice that we used `abs()` — a built-in function that returns the **absolute value** (always positive) of a number. You can check what any built-in function does with `help()`:

````python
help(abs)  # abs(-7) → 7, abs(3) → 3
````

> **Important — Indentation matters!** In Python, indentation isn't just for readability — it's part of the language. Getting it wrong will cause errors. Use **4 spaces** per level (most editors do this automatically). Avoid mixing tabs and spaces.

## 2. Conditional Statements — `if`, `elif`, `else`

Conditionals (also called **branching statements**) let your program **choose which code to run** based on whether a condition is `True` or `False`.

### The Structure

```python
if <condition>:
    # runs if condition is True
elif <another condition>:
    # runs if the first was False and this one is True
else:
    # runs if ALL conditions above were False
```

| Part | Required? | What it does |
|---|---|---|
| `if` | Yes | Checks the first condition |
| `elif` | Optional (can have many) | Checks additional conditions if the previous ones were `False` |
| `else` | Optional (at most one) | Catches everything that didn't match above |

Think of it like a **decision tree**: Python checks each condition from top to bottom and runs **only the first one that is `True`**. If none are true, the `else` block runs (if there is one).

### Example: Building Our Own `abs()` Function

Let's write a function that computes absolute value using `if` / `elif` / `else`:

Notice how the function has **three** possible paths — only one will run for any given input:

````python
def absolute_value(x):
    """Compute abs(x)."""
    if x > 0:        # Is x positive?
        return x      # Yes → return it as-is
    elif x == 0:      # Is x exactly zero?
        return 0      # Yes → return 0
    else:             # x must be negative
        return -x     # Flip the sign to make it positive
````

Let's test it with different inputs to see each branch in action:

````python
absolute_value(-5)  # Negative → hits the "else" branch → returns -(-5) = 5
````

````python
absolute_value(6)  # Positive → hits the "if" branch → returns 6
````

````python
absolute_value(0)  # Zero → hits the "elif" branch → returns 0
````

````python
absolute_value(-0.1)  # Negative float → "else" branch → returns 0.1
````

Here's a visual summary of the flow for `absolute_value(-5)`:

```
absolute_value(-5)
    │
    ├─ if -5 > 0?  → False (skip)
    │
    ├─ elif -5 == 0?  → False (skip)
    │
    └─ else  → runs! return -(-5) = 5
```

### Simplifying with `not`

Sometimes you can write the same logic differently. The `not` keyword **flips** a boolean: `not True` becomes `False` and vice versa.

Here's the same `absolute_value` function written more concisely — since we only care whether `x` is negative or not, we don't need `elif`:

Both versions produce the same result — the second is just shorter because `0` and positive numbers are handled together:

````python
def absolute_value(x):
    """Compute abs(x)."""
    if not x < 0:  # "not x < 0" means "x is zero or positive"
        return x
    else:           # x must be negative
        return -x
````

````python
absolute_value(-5)  # Still works the same!
````

````python
absolute_value(0)  # 0 is "not negative", so the if branch handles it
````

> **Tip:** When you find yourself writing an `elif` for a condition that's just the opposite of the `if`, you probably don't need `elif` at all — just use `else`.

## Quick Reference

### Conditional Statement Syntax

```python
if condition:
    # runs when condition is True
elif other_condition:
    # runs when first was False but this is True
else:
    # runs when everything above was False
```

### Key Rules

| Rule | Detail |
|---|---|
| Only **one** branch runs | Python picks the first `True` condition and skips the rest |
| `elif` is optional | You can have zero or many `elif` blocks |
| `else` is optional | Catches anything not matched above |
| Each header ends with `:` | Forgetting the colon is a common syntax error |
| Body must be indented | 4 spaces is the standard |

### Boolean Operators You Can Use in Conditions

| Operator | Meaning | Example |
|---|---|---|
| `<`, `>`, `<=`, `>=` | Comparisons | `x > 0` |
| `==`, `!=` | Equal / not equal | `x == 0` |
| `and` | Both must be True | `x > 0 and x < 10` |
| `or` | At least one must be True | `x < 0 or x > 100` |
| `not` | Flips True/False | `not x < 0` |
