## 1. Defining a Function

A **function** is a reusable block of code that takes some input, does something with it, and (optionally) gives back a result.

You create a function using the **`def`** keyword:

```python
def function_name(parameter1, parameter2):
    # body of the function (indented)
    return result
```

| Part | What it means |
|---|---|
| `def` | Tells Python "I'm defining a new function" |
| `function_name` | The name you choose for your function |
| `(parameter1, ...)` | The inputs the function expects (also called **parameters**) |
| `:` | Marks the start of the function body |
| Indented lines | The code that runs when the function is called |
| `return` | Sends a value back to whoever called the function |

Let's create a simple function that squares a number:

**Important:** The body of the function **must be indented** (usually 4 spaces). This is how Python knows which lines belong to the function.

````python
def square(x):
    return x * x  # Multiply x by itself and return the result
````

Running the cell above **defines** the function but doesn't run it yet. Nothing happens until we **call** it — just like with built-in functions, we use the name followed by parentheses with an argument:

````python
v = square(5)  # Call square with 5 → returns 5 * 5 = 25, stored in v
````

````python
v  # The value that square(5) returned
````

Let's break down exactly what happened:

1. We **defined** `square` — Python remembers the function but doesn't run it
2. We **called** `square(5)` — Python jumps into the function, sets `x = 5`, computes `5 * 5 = 25`, and **returns** `25`
3. The returned value (`25`) gets stored in the variable `v`

> **Key idea:** `return` is what makes the function give back a value. Without it, the function would do its work but produce `None` (Python's way of saying "nothing").

## 2. Calling Functions Inside Functions

Functions become really powerful when you **use them as building blocks**. You can call one function from inside another to build up more complex behavior — just like snapping LEGO pieces together.

Let's write a function that uses our `square()` function to compute the **sum of two squares** (x² + y²):

````python
def sum_squares(x, y):
    return square(x) + square(y)  # Calls square() twice, then adds the results
````

````python
result = sum_squares(3, 4)  # square(3) + square(4) = 9 + 16 = 25
````

````python
result  # 3² + 4² = 9 + 16 = 25
````

Here's the step-by-step breakdown of `sum_squares(3, 4)`:

1. Python calls `sum_squares` with `x = 3` and `y = 4`
2. Inside `sum_squares`, it calls `square(3)` → returns `9`
3. Then it calls `square(4)` → returns `16`
4. It adds the results: `9 + 16 = 25`
5. Returns `25`

### What Happens with Wrong Types?

Our `square` function uses `*`, which works on numbers. But what if we pass something that can't be multiplied? Python will raise a **TypeError**:

````python
sum_squares('a', 3)  # 'a' * 'a' doesn't make sense — this will error!
````

> **Takeaway:** Python doesn't check the type of arguments when you *define* a function — it only checks when you *call* it. If the operations inside don't make sense for the given types, you'll get an error at that point.

## 3. Default Parameter Values

Sometimes a parameter has a value that makes sense *most of the time*. Instead of forcing the caller to provide it every time, you can give it a **default value**.

If the caller doesn't provide that argument, Python uses the default. If they do provide it, their value overrides the default.

### The Problem — Without Defaults

Let's define a simple `sqrt` function that takes a number and an initial guess. Both arguments are required:

**Note:** This is a simplified example to demonstrate the concept — it just prints a message rather than actually computing the square root.

````python
def sqrt(num, guess):
    print("Finding square root of", num, "starting with guess:", guess)
````

````python
sqrt(36, 1)  # Works fine — we provided both arguments
````

But what if we skip the `guess` argument? Since both parameters are required, Python will complain:

````python
sqrt(36)  # Error! We didn't provide the required "guess" argument
````

### The Solution — With a Default Value

We can fix this by giving `guess` a **default value** using `=` in the parameter list. Now callers *can* provide a guess, but they don't *have to*:

````python
def sqrt(num, guess=1.0):  # guess defaults to 1.0 if not provided
    print("Finding square root of", num, "starting with guess:", guess)
````

````python
sqrt(36)  # We only passed num — guess automatically uses the default (1.0)
````

````python
sqrt(25, 3)  # We provided both — our value (3) overrides the default
````
