```python
def FireAlarmSystem(smoke, hightempt):
    
    if smoke:
        return {
            "Alarm": True, 
            "Sprinkler": True, 
            "Fire Department": False
        }
        
    elif hightempt:
        return {
            "Alarm": True, 
            "Sprinkler": False, 
            "Fire Department": True
        }
        
    else:
        return {
            "Alarm": False, 
            "Sprinkler": False, 
            "Fire Department": False
        }

print("Fire Alarm System")

tests = [
    ("No Smoke", False, False),
    ("Smoke detected", True, False),
    ("High temperature", False, True),
    ("Both smoke and high tempt", True, True)
]

for test_name, smoke, tempt in tests:
    
    result = FireAlarmSystem(smoke, tempt)
    
    print(f"\n\nTest: {test_name}")
    print(f"\nInput - Smoke: {smoke}, High Temp: {tempt}")
    print(f"Output - Alarm: {result['Alarm']}, Sprinkler: {result['Sprinkler']}, Fire Dept: {result['Fire Department']}")
```

```python
Dry_threshold = 25
Moist_threshold = 50


def watering_agent(moisture_level):
    
    if moisture_level < Dry_threshold:
        return "Dry soil", "Activate"
    
    elif moisture_level <= Moist_threshold:
        return "Moist soil", "Off"
    
    else:
        return "Wet soil", "Deactivate"


print("\nAgent Rules:\n")

print(f"1. If moisture < {Dry_threshold}% (Dry soil)")
print("Then: activate watering system")

print(f"\n2. If moisture {Dry_threshold}-{Moist_threshold}% (Moist soil)")
print("Then: keep watering system off")

print(f"\n3. If moisture > {Moist_threshold}% (Wet soil)")
print("Then: Deactivate watering system")

print("\nTesting agent:\n")

test_cases = [
    (15, "Dry soil, water needed"),
    (45, "Moist soil, nothing needed"),
    (75, "Wet soil, too much water"),
    (25, "Dry soil, water needed"),
    (60, "Moist soil, nothing needed"),
    (10, "Dry soil, water needed"),
    (50, "Moist soil, nothing needed"),
    (85, "Wet soil, too much water")
]

for moisture, description in test_cases:
    
    print(f"\nTest: {description}")
    print(f"Moisture level: {moisture}%")
    
    state, action = watering_agent(moisture)
    
    print(f"Agent detects: {state}")
    print(f"Agent action: {action} watering system")

    
    if action == "Activate":
        print("Watering activated")
        
    elif action == "Deactivate":
        print("Watering deactivated")
        
    else:
        print("No action needed")

        
print("Error handling demenostration")


def moisture_check():
    
    test_inputs = ["25", "-0", "-5", "150", "45.5"]
    
    for input_val in test_inputs:
        
        print(f"\nInput: {input_val}")
        
        try:
            moisture = float(input_val)
            
            if moisture < 0 or moisture > 100:
                print(f"Error: Moisture {moisture}% is invalid (must be 0-100%)")
                
            else:
                state, action = watering_agent(moisture)
                print(f"Valid: Moisture:{moisture}%, State:{state}, Action:{action}")
                
        except ValueError:
            print(f"Error: '{input_val}' is not a valid number")

moisture_check()

print("\n\nloging Demonestration")


import time


def log_action(moisture, state, action):
    
    timestamp = time.strftime("%H:%M:%S")
    
    print(f"[{timestamp}] Sensor: {moisture}% | State: {state} | Action: {action}")

print("\nSimulating 5 sensor readings with logging:\n")


readings = [20, 40, 65, 15, 55]

for reading in readings:
    state, action = watering_agent(reading)
    log_action(reading, state, action)
    time.sleep(0.5)

```
