```python
class Node:
    def __init__(self, city, parent=None, g=0, h=0):
        self.city = city
        self.parent = parent
        self.g = g
        self.h = h
        self.f = g + h
```

```python
import math
```

```python
def heuristic(city1, city2):
    lat1, lon1 = lat_long[city1]
    lat2, lon2 = lat_long[city2]
    return math.sqrt((lat1 - lat2)**2 + (lon1 - lon2)**2) * 111
```

```python
def astar_search(graph, start, goal):

    start_node = Node(city=start, parent=None, g=0, h=heuristic(start, goal))
    
    open_list = [start_node]
    closed_set = set()

    while open_list:
        
        open_list.sort(key=lambda node: node.f)
        current = open_list.pop(0)
        
        if current.city == goal:
            path = []
            node = current
            
            while node:
                path.append(node.city)
                node = node.parent
                
            path.reverse()
            return path, current.g
            
        if current.city in closed_set:
            continue
            
        closed_set.add(current.city)
        
        for neighbor, cost in graph.get(current.city, []):
            if neighbor not in closed_set:
                g_new = current.g + cost
                h_new = heuristic(neighbor, goal)
                child = Node(city=neighbor, parent=current, g=g_new, h=h_new)
                open_list.append(child)

    print("No path found")
    return [], 0
```

```python
path, distance = astar_search(romania_map, 'Arad', 'Bucharest')

print("Shortest path from Arad to Bucharest:", path)
print("Distance:", distance)
```
