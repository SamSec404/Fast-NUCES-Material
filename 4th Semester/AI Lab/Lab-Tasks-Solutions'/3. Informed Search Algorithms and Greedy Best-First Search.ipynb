```python
class Node:
    def __init__(self, city, parent=None, g=0, h=0):
        self.city = city
        self.parent = parent
        self.g = g
        self.h = h
```

```python
import heapq
```

```python
def best_first_search(graph, start, goal='Bucharest'):

    start_node = Node(city=start, parent=None, g=0, h=heuristic_values[start])
    
    open_list = [start_node]
    closed_set = set()

    while open_list:
        open_list.sort(key=lambda node: node.h)
        current = open_list.pop(0)

        if current.city in closed_set:
            continue
            
        closed_set.add(current.city)

        if current.city == goal:
            path = []
            node = current
            
            while node:
                path.append(node.city)
                node = node.parent
                
            path.reverse()
            return path, current.g
            
        for neighbor, cost in graph.get(current.city, []):
            if neighbor not in closed_set:
                g_new = current.g + cost
                h_new = heuristic_values[neighbor]
                child = Node(city=neighbor, parent=current, g=g_new, h=h_new)
                open_list.append(child)

    print("No path found!")
    return [], 0
```

```python
path, distance = best_first_search(romania_map, 'Arad')
print("Path:", path)
print("Distance:", distance)
```
